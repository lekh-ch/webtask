# FEATURE ENGINEERING

```python
df['OrderDate']=pd.to_datetime(df['OrderDate'])
df['Year']=df['OrderDate'].dt.year

df['MonthNo']=df['OrderDate'].dt.month

df['Month'] = df['OrderDate'].dt.month_name()

df['Quarter']=df['OrderDate'].dt.quarter

df['WeekDay']=df['OrderDate'].dt.day_name()

```

```python
df['SalesBeforeDiscount']=df['Quantity'] * df['UnitPrice']

df['DiscountAmount']=df['SalesBeforeDiscount'] * (df['Discount']/100)

df['NetSales']=df['SalesBeforeDiscount'] - df['DiscountAmount']

df['ApproxProfit']=df['NetSales']-df['ShippingCost']-df['Tax']
```

```python
df.head()
```

```python
print("Total Revenue : $",round(df["TotalAmount"].sum(),2))

print("Total Orders :",df["OrderID"].nunique())

print("Total Customers :",df["CustomerID"].nunique())

print("Total Products :",df["ProductID"].nunique())

print("Total Sellers :",df["SellerID"].nunique())

print("Average Order Value : $",round(df["TotalAmount"].mean(),2))

print("Average Quantity :",round(df["Quantity"].mean(),2))

print("Average Shipping : $",round(df["ShippingCost"].mean(),2))

print("Average Discount :",round(df["Discount"].mean(),2))
```
