```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("ggplot")

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("cleaned_amazon_orders.csv")
```

```python
df['OrderDate']=pd.to_datetime(df['OrderDate'])
```

```python
df.head()
df.info()
df.columns
df.shape
```

```python
print("========== AMAZON SALES KPI SUMMARY ==========\n")

print("Total Revenue           : $", round(df["TotalAmount"].sum(),2))
print("Total Orders            :", df["OrderID"].nunique())
print("Total Customers         :", df["CustomerID"].nunique())
print("Total Products          :", df["ProductID"].nunique())
print("Total Sellers           :", df["SellerID"].nunique())
print("Average Order Value     : $", round(df["TotalAmount"].mean(),2))
print("Average Quantity        :", round(df["Quantity"].mean(),2))
print("Average Discount (%)    :", round(df["Discount"].mean(),2))
print("Average Shipping Cost   : $", round(df["ShippingCost"].mean(),2))
print("Average Tax             : $", round(df["Tax"].mean(),2))
```

```python
monthly_sales = (df.groupby('Month')['TotalAmount']
                 .sum()
                 .reset_index())
```

```python
month_order=['January','February','March','April','May','June','July','August','September','October','November','December']
monthly_sales['Month']=pd.Categorical(monthly_sales['Month'],categories=month_order,ordered=True)

monthly_sales = monthly_sales.sort_values('Month')
monthly_sales
```

```python
year_sales=(df.groupby('Year')['TotalAmount']
            .sum()
            .reset_index()
            )
```

```python
quarter_sales= (df.groupby('Quarter')['TotalAmount']
                .sum()
                .reset_index())
```

```python
category_sales = (df.groupby('Category')
.agg(
    Revenue = ('TotalAmount', 'sum'),
    Orders = ('OrderID', 'count'),
    Quantity = ('Quantity', 'sum')
)
.sort_values(by='Revenue',ascending=False)
)
category_sales
```

```python
high_yr_sales=year_sales.idxmax()
high_yr_rev=year_sales.max()
high_yr_sales
high_yr_rev
```

```python
low_mon_sales=monthly_sales.idxmin()
low_mon_rev=monthly_sales.min()
low_mon_sales
low_mon_rev
```
